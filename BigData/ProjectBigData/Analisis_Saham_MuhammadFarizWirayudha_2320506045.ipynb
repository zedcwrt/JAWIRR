```python
import pandas as pd
import pandas as lag_plot
import matplotlib.pyplot as plt
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error
from pandas.plotting import lag_plot

df = pd.read_csv('/content/dataset/Stocks/eex.us.txt')
df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Membagi data menjadi train dan test
train_data, test_data = df[0:int(len(df) * 0.8)], df[int(len(df) * 0.8):]

plt.figure(figsize=(12, 7))

# Judul dan label sumbu
plt.title('Stock over Prices')
plt.xlabel('Dates')
plt.ylabel('Prices')

# Plot data pelatihan
plt.plot(train_data['Open'], color='blue', label='Training Data')

# Plot data pengujian
plt.plot(test_data['Open'], color='green', label='Testing Data')

# Atur xticks secara dinamis
x_tick_locations = np.arange(0, len(df), 300)  # Lokasi xticks berdasarkan panjang data
x_tick_labels = df['Date'][0:len(df):300]      # Label sesuai lokasi

plt.xticks(x_tick_locations, x_tick_labels, rotation=45)  # Rotasi untuk kejelasan

# Tambahkan legenda
plt.legend()

# Tampilkan plot
plt.show()
```

```python
def smape_kun(y_true, y_pred):
    return np.mean((np.abs(y_pred - y_true) * 200/ (np.abs(y_pred) + np.abs(y_true))))
```

```python
train_ar = train_data['Open'].values
test_ar = test_data['Open'].values

history = [x for x in train_ar]
print(type(history))
predictions = list()
for t in range(len(test_ar)):
    model = ARIMA(history, order=(5,1,0))
    model_fit = model.fit()
    output = model_fit.forecast()
    yhat = output[0]
    predictions.append(yhat)
    obs = test_ar[t]
    history.append(obs)
    #print('predicted=%f, expected=%f' % (yhat, obs))
error = mean_squared_error(test_ar, predictions)
print('Testing Mean Squared Error: %.3f' % error)
error2 = smape_kun(test_ar, predictions)
print('Symmetric mean absolute percentage error: %.3f' % error2)
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(12, 7))

# Plot data pelatihan
plt.plot(df['Open'], color='blue', label='Training Data')

# Plot prediksi
aligned_index = test_data.index[:len(predictions)]  # Sesuaikan indeks prediksi dengan panjang data
plt.plot(aligned_index, predictions, color='green', marker='o', linestyle='dashed', label='Predicted Price')

# Plot data aktual
plt.plot(test_data.index, test_data['Open'], color='red', label='Actual Price')

# Judul dan label sumbu
plt.title('Prices Prediction')
plt.xlabel('Dates')
plt.ylabel('Prices')

# Atur xticks
x_tick_locations = np.arange(0, len(df), 300)  # Lokasi xticks berdasarkan panjang data
x_tick_labels = df['Date'][0:len(df):300]      # Label xticks sesuai lokasi
plt.xticks(x_tick_locations, x_tick_labels, rotation=45)  # Rotasi untuk kejelasan

# Tambahkan legenda
plt.legend()

# Tampilkan plot
plt.show()
```
